 Calculate the average of 3.14, 2.718, 1.618, 0.577

```python
numbers=[3.14,2.718,1.618,0.577]
average=sum(numbers)/len(numbers)
print("average:",average)
```

 2: Convert 98.6 Fahrenheit to Celsius (F = C * 9/5 + 32)

```python
fahrenheit=98.6
celsius =(fahrenheit-32)*5/9
print("temperature in celsius:",celsius)
```

3: Calculate compound interest on $1000 at 5.5% for 3 years"

```python
principle=1000
rate=5.5
time=3
amount=principle *(1+rate/100)**time
compound_interest=amount-principle
print("compound interest:",round(compound_interest,2))
print("total amount:",round(amount,2))
```

4 Find the hypotenuse of a right triangle with sides 3.5 and 4.2

```python
import math
a=3.5
b=4.2
hypotenuse=math.sqrt(a**2+b**2)
print("hypotenuse:",round(hypotenuse,2))
```

5: Calculate the volume of a sphere with radius 7.8"

```python
import math
radius=7.8
volume=(4/3)*math.pi*radius**3
print("volume of sphere:",round(volume,2))
```
