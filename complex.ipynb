1: Create complex number 5 + 3j

```python
a=5+3j
print(a)
print(type(a))
```

2: Find the real part of complex number 7 - 2j

```python
a=7-2j
print(a.real)
print(type(a))
```

3: Find the imaginary part of complex number 4 + 6j

```python
a=4+6j
print(a.imag)
print(type(a))
```

4: Add two complex numbers: (3 + 2j) and (1 + 4j)

```python
a=3+2j
b=1+4j
c= a + b
print(c)
print(type(c))
```

5: Multiply two complex numbers: (2 + 3j) and (1 + 2j)

```python
a=2+3j
b=1+2j
c= a * b
print(c)
print(type(c))
```

 6: Find the magnitude of complex number 6 + 8j

```python
a=6+8j
print(abs(a))
```

7: Find the conjugate of complex number 5 - 7j

```python
a=5-7j
print(a.conjugate())
print(type(a))
```

8: Subtract complex numbers: (10 + 5j) - (3 + 2j)

```python
a=10+5j
b=3+2j
c=a - b
print(c)
print(type(c))
```

9: Divide complex numbers: (8 + 6j) / (2 + 1j)

```python
a=8+6j
b=2+1j
c= a / b
print(c)
print(type(c))
```

10: Find the phase angle of complex number 1 + 1j

```python
import cmath
a=1 + 1j
print(cmath.phase(a))
print(type(a))
```
