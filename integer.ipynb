5 Calculate the sum of digits in number 12345

```python
number=12345
total=0
while number > 0:
    digit=number%10
    total +=digit
    number //=10
    print(total)
```

6 Check if 97 is a prime number

```python
num=97
if num >1:
    for i in range(2,num):
        if(num%i)==0:
            print("print not a prime number")
            break
        else:
            print(" prime number")
            break
    else:
        print("not a prime number")
```

7 Find the factorial of 8

```python
num=8
factorial=1
for i in range(1,num+1):
    factorial *=i
    print(factorial)
```
